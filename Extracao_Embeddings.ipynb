```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
# Lê o arquivo gerado no notebook anterior

caminho_arquivo = '/content/drive/MyDrive/TCC/base_pronta_para_extracao_embeddings.csv'
dataset_final = pd.read_csv(caminho_arquivo)
```

```python
print("Carregando o modelo BGE-m3...")
print("(Como é a primeira vez, o download de ~2GB iniciará. Pode levar alguns minutos!)")
modelo_bge = SentenceTransformer('BAAI/bge-m3')
```

```python
coluna_alvo = 'text'
```

```python
print("Extraindo embeddings avançados (Processando em lote)...")
# Transforma a coluna 'text' em uma lista do Python
# O TRUQUE DE MESTRE: O .astype(str) obriga o Pandas a transformar qualquer
# valor nulo/decimal (NaN) em um texto vazio comum, evitando que o modelo quebre.
textos_extraidos = dataset_final[coluna_alvo].astype(str).tolist()

# O comando encode transforma o texto em matemática.
embeddings = modelo_bge.encode(textos_extraidos, show_progress_bar=True)

# O modelo BGE-m3 gerou uma matriz matemática gigante separada da base de dados original.
# O comando list(embeddings) pega todos esses vetores numéricos e cria uma nova coluna chamada vetores no dataset_final.
dataset_final['vetores'] = list(embeddings)
print("Textos transformados em matemática com sucesso!")
```

```python
# Salva o dataset inteiro mantendo a matriz matemática intacta
dataset_final.to_parquet("base_final_com_vetores.parquet")
print("Tabela salva com segurança no formato Parquet!")
```

```python
# Salva a tabela inteira (textos + vetores) de forma blindada no Drive
caminho_parquet = '/content/drive/MyDrive/TCC/base_final_com_vetores.parquet'
dataset_final.to_parquet(caminho_parquet)

print(f"Tabela salva com segurança em: {caminho_parquet}")
```
